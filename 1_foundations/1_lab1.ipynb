```python
# Check that your Google API key loaded from the project-root .env file

import os
google_api_key = os.getenv("GOOGLE_API_KEY")

if google_api_key:
    print("Google API key is set")
else:
    print("Google API key not set. Add GOOGLE_API_KEY to .env and save the file.")
```

```python
# The OpenAI SDK is compatible with Google's Gemini API endpoint

from openai import OpenAI
```

```python
# Configure the OpenAI-compatible client to use Google's Gemini API

google_api_key = os.getenv("GOOGLE_API_KEY")
if not google_api_key:
    raise ValueError("GOOGLE_API_KEY is not set. Add it to the project-root .env file and run the dotenv cell.")

gemini = OpenAI(
    api_key=google_api_key,
    base_url="https://generativelanguage.googleapis.com/v1beta/openai/",
)
model = "gemini-2.5-flash"
```

```python
# Create a list of messages in the familiar OpenAI format

messages = [{"role": "user", "content": "Tell me a fun fact"}]
```

```python
messages
```

```python
# Call Gemini through Google's OpenAI-compatible endpoint

response = gemini.chat.completions.create(model=model, messages=messages)
print(response.choices[0].message.content)
```

```python
# And now - let's ask for a question:

question = "Please propose a hard, challenging question to assess someone's IQ. Respond only with the question."
messages = [{"role": "user", "content": question}]
```

```python
# Ask Gemini to propose a challenging question

response = gemini.chat.completions.create(model=model, messages=messages)
question = response.choices[0].message.content
print(question)
```

```python
# form a new messages list

messages = [{"role": "user", "content": question}]
messages
```

```python
# Ask Gemini to answer the question

response = gemini.chat.completions.create(model=model, messages=messages)
answer = response.choices[0].message.content
print(answer)
```

```python
from IPython.display import Markdown, display

display(Markdown(answer))
```

```python
message = f"""
Here is a question:
{question}

And here is a possible answer that might be correct or incorrect:
{answer}

Please evaluate if the answer is correct or incorrect.
"""

print(message)
```

```python
messages = [{"role": "user", "content": message}]
response = gemini.chat.completions.create(model=model, messages=messages)
print(response.choices[0].message.content)
```
